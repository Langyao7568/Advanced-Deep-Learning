## 2. Lists, indexing, and unpacking

A **list** stores an ordered sequence of values. Python indices start at **0**, so the second item has index 1. Negative indices count from the end. A slice `[start:stop]` includes `start` but excludes `stop`; `[:stop]` starts at the beginning.

Lists can be changed, and `.append(value)` adds an item. These operations will be useful for storing learning curves. Conversely, a **tuple**, such as `(2,3)`, is an ordered sequence whose entries cannot be replaced. You will often see tuples used for shapes and multiple return values.


```python
values = [2, 4, 6, 8] # this is a list
print('First value:', values[0], 'Last value:', values[-1])
print('Middle slice:', values[1:3])
values.append(10)
print('After append:', values, 'Length:', len(values))

point = (3, 4) # this is a tuple
# point[0] = 5  # this will raise an error because tuples are immutable
x, y = point  # unpack two values into two variables
print('Point coordinates:', x, y)
```

## 3. Conditions and loops

`if`, `elif`, and `else` choose which code to execute. A colon starts a block; **indentation defines the block**, usually with four spaces.

A `for` loop visits each item in a sequence. `range(3)` gives 0, 1, and 2; its upper bound is excluded. `enumerate(values)` provides both index and value, while `zip(a,b)` pairs corresponding items from two sequences. Tuple unpacking lets us assign each pair to two variables.


```python
score = 0.75
if score >= 0.8: # this is a conditional statement
    print('High score')
elif score >= 0.5: # elif is short for "else if"
    print('Medium score')
else:
    print('Low score')

for step in range(3): # loop over 0, 1, 2
    print('Step:', step)

for index, loss in enumerate(losses): # enumerate gives both the index and the value
    print(f'Entry {index}: loss {loss:.2f}')

for label, value in zip(['x', 'y'], [3, 4]): # zip pairs up elements from two lists (like a zipper)
    print(label, value)
```

## 4. Functions

Use `def` to define a reusable function. Its arguments are the inputs; `return` passes a result back to the caller. An argument can have a default value. Keyword arguments make calls explicit, as in `scale(3,factor=0.5)`.

The text just below the function definition is its **docstring**. Returning several values creates a tuple, which we can unpack.


```python
def scale(value, factor=2): # define a function with a default argument 
    """Multiply a value by a factor."""
    return value * factor

print('Default factor:', scale(3)) # no factor specified, so it uses the default value of 2
print('Specified factor:', scale(3, factor=0.5))

def summarize(values): 
    """Return the count and mean of a non-empty sequence."""
    return len(values), sum(values) / len(values) # return multiple values as a tuple

count, mean_value = summarize([2, 4, 6]) # unpack the returned tuple into two variables
print('Count:', count, 'Mean:', mean_value) 
```

## 5. NumPy arrays and shapes

A NumPy **array** stores numerical data. A vector is a one-dimensional array; a matrix has two dimensions. `.shape` is a tuple of dimension lengths, `.ndim` is the number of dimensions, and `.dtype` is the stored numerical type.

We will normally store one sample per row. An array of shape `(3,2)` then represents three samples with two features each. Floating-point inputs and integer class labels will also be important in PyTorch.

`np.array` creates an array from a list. `np.arange(n)` creates values from 0 to n−1. `np.zeros(shape)` creates an array filled with zeros.


```python
vector = np.array([1, 2, 3], dtype=np.float32) # create a 1D array (vector) of floats
features = np.array([[1, 2], [3, 4], [5, 6]], dtype=np.float32) # create a 2D array (matrix) of floats
labels = np.array([0, 1, 0], dtype=np.int64) # create a 1D array (vector) of integers
print('Vector:', vector, 'Shape:', vector.shape) 
print('Features:\n', features)
print('Feature shape:', features.shape, 'Dimensions:', features.ndim, 'Type:', features.dtype)
print('Labels:', labels, 'Shape:', labels.shape, 'Type:', labels.dtype)
print('Sequence:', np.arange(5))
print('Zeros:\n', np.zeros((2, 3), dtype=np.float32))
```

### Select values, rows, and columns

Use `a[row,column]` for a matrix entry. `a[0]` or `a[0,:]` selects the first row; `a[:,0]` selects the first column. Slices use the same rules as for lists.

A comparison such as `labels == 0` produces a boolean array. It can be used as a **mask** to select matching rows. Basic slicing can share data with the original array; use `.copy()` if you need an independent array to modify.


```python
print('Entry in row 1, column 0:', features[1, 0])
print('First row:', features[0, :])
print('First column:', features[:, 0])
print('First two rows:\n', features[:2, :])
mask = labels == 0 # create a boolean mask for samples with label 0
print('Mask:', mask)
print('Samples with label 0:\n', features[mask]) # apply the mask to select rows where the label is 0
```

## 6. Array calculations: elementwise operations and matrix multiplication

Lists and arrays behave differently. Multiplying a list by 2 repeats the list; multiplying an array by 2 multiplies each value. For arrays, `+`, `-`, `*`, `/`, and `**` work **elementwise**. NumPy functions such as `np.exp` also operate on every entry, so our sigmoid function works on a whole array.

For matrix multiplication, use **`@`**, not `*`. An `(N,D)` input matrix multiplied by a `(D,H)` weight matrix gives an `(N,H)` result. The two inner dimensions must match. `.T` transposes a matrix.


```python
print('List multiplied by 2:', [1, 2, 3] * 2)
print('Array multiplied by 2:', vector * 2)
print('Elementwise square:', vector ** 2)
print('Sigmoid applied to an array:', sigmoid(np.array([-1.0, 0.0, 1.0])))

A = np.array([[1, 2], [3, 4]], dtype=np.float32)
B = np.array([[5, 6], [7, 8]], dtype=np.float32)
print('Elementwise multiplication:\n', A * B)
print('Matrix multiplication:\n', A @ B)
print('Transposed A:\n', A.T)
```

### Broadcasting and reductions

**Broadcasting** applies a smaller array across a compatible larger array. Adding a vector of shape `(2,)` to `features` of shape `(3,2)` adds the same two values to every row. This will later let us add a bias to every sample in a batch.

A reduction such as `.mean()` combines values. For a matrix, `axis=0` combines rows and returns one value per column; `axis=1` combines columns and returns one value per row. `keepdims=True` retains the reduced dimension with length 1.


```python
offset = np.array([10, 100], dtype=np.float32)
print('Offset added to every row:\n', features + offset)
print('Mean of all entries:', features.mean())
print('Mean per column:', features.mean(axis=0))
print('Mean per row:', features.mean(axis=1))
print('Mean per row, keeping the dimension:\n', features.mean(axis=1, keepdims=True))
```

## 7. Reshaping a small batch of images

`.reshape(...)` changes the arrangement of an array without changing its number of entries. One dimension can be `-1`; NumPy calculates that dimension from the total size.

For image classification, we often flatten each image into a feature vector. Keep the first dimension, which counts the images (this is called a **batch** of images). The tiny 2×3 images below are just an example; the same idea works for larger images and PyTorch tensors.

A one-dimensional vector has shape `(D,)`. A matrix with one row has shape `(1,D)` and a matrix with one column has shape `(D,1)`. These shapes behave differently in matrix operations.


```python
print('Vector shape:', vector.shape) # this is a 1D array with 3 elements; it has no explicit row or column dimension, so it is neither a row vector nor a column vector
print('One row:', vector.reshape(1, -1).shape)
print('One column:', vector.reshape(-1, 1).shape)

images = np.arange(12, dtype=np.float32).reshape(2, 2, 3)
print('Two images:\n', images)
print('Image batch shape:', images.shape)
```

## 8. Reading a simple class

A **class** groups values and functions into an object. `__init__` initializes the object; `self` refers to that object. Stored values such as `self.factor` are **attributes**; a function such as `.apply(...)` is a **method**.


```python
class Scaler: 
    def __init__(self, factor):
        self.factor = factor

    def apply(self, values):
        return values * self.factor

scaler = Scaler(factor=0.5)
print('Stored factor:', scaler.factor)
print('Scaled array:', scaler.apply(vector))
```
